# TF-IDF baseline

Synthetic, offline educational experiment. No live mailbox is accessed.


```python
from sklearn.feature_extraction.text import TfidfVectorizer
corpus=['meeting request tomorrow', 'lottery winning payment', 'data extortion crypto wallet']
v=TfidfVectorizer(ngram_range=(1,2))
x=v.fit_transform(corpus)
print(x.shape,v.get_feature_names_out().tolist())
```
