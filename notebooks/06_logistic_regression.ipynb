# Logistic-regression baseline

Synthetic, offline educational experiment. No live mailbox is accessed.


```python
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
x=['normal project meeting','send crypto payment now','client agenda notes','you won lottery wire transfer','team planning meeting','stolen data pay crypto']
y=[0,1,0,1,0,1]
model=make_pipeline(TfidfVectorizer(),LogisticRegression(max_iter=500))
model.fit(x,y)
print(model.predict(['wire transfer prize','team meeting']))
```
